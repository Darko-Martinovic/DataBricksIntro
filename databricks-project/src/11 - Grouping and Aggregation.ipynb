📌 Creating a Sample DataFrame

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
from pyspark.sql import functions as F

schema = StructType([
    StructField("employee_id", IntegerType(), True),
    StructField("name",        StringType(),  True),
    StructField("department",  StringType(),  True),
    StructField("salary",      IntegerType(), True),
    StructField("years",       IntegerType(), True)
])

data = [
    (1,  "Alice",   "Engineering", 95000, 5),
    (2,  "Bob",     "Marketing",   62000, 3),
    (3,  "Carol",   "Engineering", 87000, 8),
    (4,  "David",   "HR",          54000, 2),
    (5,  "Eve",     "Marketing",   71000, 6),
    (6,  "Frank",   "Engineering", 110000, 10),
    (7,  "Grace",   "HR",          58000, 4),
    (8,  "Henry",   "Marketing",   67000, 1),
    (9,  "Isla",    "Engineering", 92000, 7),
    (10, "Jack",    "HR",          60000, 3)
]

df = spark.createDataFrame(data, schema=schema)
df.show()
```

📌 groupBy() with count()

```python
# Count the number of employees per department
df.groupBy("department").count().show()
```

📌 groupBy() with sum() and avg()

```python
# Total and average salary per department
df.groupBy("department").agg(
    F.sum("salary").alias("total_salary"),
    F.avg("salary").alias("avg_salary")
).show()
```

📌 groupBy() with min() and max()

```python
# Salary range and headcount per department
df.groupBy("department").agg(
    F.min("salary").alias("min_salary"),
    F.max("salary").alias("max_salary"),
    F.count("employee_id").alias("headcount")
).show()
```

📌 Multiple Aggregations with agg()

```python
# Combined summary: count, total salary, average years of experience
df.groupBy("department").agg(
    F.count("employee_id").alias("headcount"),
    F.sum("salary").alias("total_salary"),
    F.round(F.avg("salary"), 2).alias("avg_salary"),
    F.round(F.avg("years"), 1).alias("avg_years")
).orderBy("department").show()
```
