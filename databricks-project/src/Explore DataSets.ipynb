# Explore DataSets

A **Dataset** in Spark is a strongly-typed, distributed collection of data. Unlike a DataFrame (which is `Dataset[Row]`), a typed Dataset lets the compiler catch type errors at compile time rather than at runtime.

| Concept | Description |
|---------|-------------|
| `case class` | Defines the schema and gives you named, typed fields |
| `toDS()` | Converts a `Seq` or `RDD` into a `Dataset[T]` |
| `spark.implicits._` | Provides the implicit encoders needed for `toDS()` / `toDF()` |
| Type-safe operations | Lambda functions on `Dataset[T]` work on the actual case class — compiler checks field names and types |

```scala
// Define a case class — this is what gives you type safety
case class Person(name: String, age: Int)

// Create sample data
val data = Seq(
  Person("Alice", 30),
  Person("Bob", 25),
  Person("Carol", 35)
)

// Create a Dataset (not just a DataFrame)
import spark.implicits._
val ds = data.toDS()

// The compiler knows ds is Dataset[Person]
// This would fail at COMPILE TIME if you got the type wrong
ds.filter(p => p.age > 28).show()
```
