# Data Join Exercise

```python
import os
if not os.getenv("DATABRICKS_RUNTIME_VERSION"):
    from spark_setup import spark, display  # local run through Databricks Connect
```

```python
df = spark.read.table("population_metrics.default.countries_population")
display(df)
```

```python
df_region = spark.read.table("population_metrics.default.country_regions")
df_subregion = spark.read.table("population_metrics.default.country_sub_regions")

df_joined = (
    df.join(df_region, df["REGION_ID"] == df_region.id, "inner")
    .join(df_subregion, df["SUB_REGION_ID"] == df_subregion.id, "inner")
    .select(
        df["COUNTRY_ID"],
        df["NAME"].alias("COUNTRY"),
        df_region["NAME"].alias("REGION"),
        df_subregion["NAME"].alias("SUB_REGION"),
        df["POPULATION"],
        df["AREA_KM2"],
    )
)

display(df_joined)
#Save df_joined as a table under population_metrics.default.countries_consolidated
df_joined.write.mode("overwrite").saveAsTable("population_metrics.default.countries_consolidated")
```
