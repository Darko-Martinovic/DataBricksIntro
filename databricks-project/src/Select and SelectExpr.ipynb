# Select and SelectExpr
Selecting and transforming columns using `select()` with `col()` expressions and `selectExpr()` with SQL string expressions.

### Links and Resources
- [select()](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrame.select.html)
- [selectExpr()](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrame.selectExpr.html)
- [col()](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.functions.col.html)

📌 Spark session setup

```python
import os
if not os.getenv("DATABRICKS_RUNTIME_VERSION"):
    from spark_setup import spark, display
```

📌 Reading data into a Spark DataFrame

```python
df = spark.read.table("population_metrics.default.countries_population")
display(df)
```

📌 select() — pick columns by name

```python
df.select("NAME", "POPULATION", "AREA_KM2").show(5)
```

📌 select() with col() — rename and derive columns

```python
from pyspark.sql.functions import col, round

df.select(
    col("NAME").alias("Country"),
    col("POPULATION"),
    round(col("AREA_KM2"), 0).alias("Area")
).show(5)
```

📌 selectExpr() — SQL string expressions (no imports needed)

```python
df.selectExpr(
    "NAME AS Country",
    "POPULATION",
    "ROUND(AREA_KM2, 0) AS Area"
).show(5)
```

📌 selectExpr() — computed columns with SQL expressions

```python
df.selectExpr(
    "NAME AS Country",
    "POPULATION",
    "AREA_KM2",
    "ROUND(POPULATION / AREA_KM2, 2) AS Population_Density"
).show(5)
```
