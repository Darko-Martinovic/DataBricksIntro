# withColumn and withColumns
Adding and transforming columns on an existing DataFrame without rewriting the full column list.

### Links and Resources
- [withColumn()](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrame.withColumn.html)
- [withColumns()](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrame.withColumns.html)
- [col()](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.functions.col.html)

📌 Spark session setup

```python
import os
if not os.getenv("DATABRICKS_RUNTIME_VERSION"):
    from spark_setup import spark, display
```

📌 Reading data into a Spark DataFrame

```python
df = spark.read.table("population_metrics.default.countries_population")
display(df)
```

📌 withColumn() — add a new computed column

```python
from pyspark.sql.functions import col, round

df.withColumn("Population_Density", round(col("POPULATION") / col("AREA_KM2"), 2)).show(5)
```

📌 withColumn() — overwrite an existing column (convert km² to m²)

```python
df.withColumn("AREA_KM2", round(col("AREA_KM2") * 1_000_000, 0)).show(5)
```

📌 withColumn() — rename a column via alias

```python
df.withColumn("Country", col("NAME")).drop("NAME").show(5)
```

📌 withColumns() — add multiple columns in one call (Spark 3.3+)

```python
from pyspark.sql.functions import upper

df.withColumns({
    "Population_Density": round(col("POPULATION") / col("AREA_KM2"), 2),
    "NAME_UPPER": upper(col("NAME"))
}).show(5)
```
