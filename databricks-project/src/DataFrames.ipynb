# DataFrames
Exploring different ways to represent data in Python before loading into Spark DataFrames.

```python
# Plain list — each element represents [name, age, city] (no labels)
data = [
    ["Alice", 30, "New York"],
    ["Bob", 25, "London"],
    ["Charlie", 35, "Paris"]
]

for row in data:
    print(row)
```

```python
# List of dictionaries — same data with explicit labels
data_labeled = [
    {"name": "Alice", "age": 30, "city": "New York"},
    {"name": "Bob", "age": 25, "city": "London"},
    {"name": "Charlie", "age": 35, "city": "Paris"}
]

for row in data_labeled:
    print(row)
```

```python
import os, sys

try:
    from databricks.connect import DatabricksSession
    spark = DatabricksSession.builder.getOrCreate()
except ImportError:
    if not os.getenv("DATABRICKS_RUNTIME_VERSION"):
        raise RuntimeError(
            f"Wrong Python kernel — 'databricks.connect' not found in:\n  {sys.executable}\n\n"
            "In VSCode: click the kernel selector (top-right of notebook) "
            "and choose the interpreter at:\n"
            r"  C:\Source\Dsa\DataBricks\.venv\Scripts\python.exe"
        )
    from pyspark.sql import SparkSession
    spark = SparkSession.builder.getOrCreate()

# DataFrame from plain list — column names must be supplied explicitly
df = spark.createDataFrame(data, schema=["name", "age", "city"])
df.show()
```

```python
# DataFrame from list of dicts — column names are inferred from the dictionary keys
df_labeled = spark.createDataFrame(data_labeled)
df_labeled.show()
```
