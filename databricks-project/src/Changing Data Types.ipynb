# Changing Data Types
Inspecting and casting column data types in a Spark DataFrame using `cast()`, `withColumn()`, and `selectExpr()`.

📌 Spark session setup

```python
import os
if not os.getenv("DATABRICKS_RUNTIME_VERSION"):
    from spark_setup import spark, display
```

📌 Reading data and inspecting the schema

```python
df = spark.read.table("population_metrics.default.countries_population")
df.printSchema()
```

📌 Inspecting column types as a flat list

```python
print(df.dtypes)
```

📌 cast() with withColumn() — change a single column type

```python
from pyspark.sql.functions import col
from pyspark.sql.types import LongType

# Cast POPULATION from IntegerType to LongType
df_cast = df.withColumn("POPULATION", col("POPULATION").cast(LongType()))
df_cast.printSchema()
```

📌 cast() using a type string — shorter syntax

```python
# Same cast using a string alias instead of a type class
df_cast = df.withColumn("POPULATION", col("POPULATION").cast("long"))
df_cast.printSchema()
```

📌 withColumns() — cast multiple columns at once

```python
df_cast = df.withColumns({
    "POPULATION": col("POPULATION").cast("long"),
    "AREA_KM2":   col("AREA_KM2").cast("integer"),
    "COUNTRY_ID": col("COUNTRY_ID").cast("string")
})

df_cast.printSchema()
```

📌 selectExpr() — cast using SQL CAST syntax

```python
df_cast = df.selectExpr(
    "CAST(COUNTRY_ID AS STRING) AS COUNTRY_ID",
    "NAME",
    "CAST(POPULATION AS LONG) AS POPULATION",
    "CAST(AREA_KM2 AS INTEGER) AS AREA_KM2"
)

df_cast.printSchema()
```

📌 Verifying the cast values in the data

```python
display(df_cast)
```
