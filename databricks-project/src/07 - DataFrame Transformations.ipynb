📌 Reading data into a Spark DataFrame

```python
import os
if not os.getenv("DATABRICKS_RUNTIME_VERSION"):
    from spark_setup import spark, display  # local run through Databricks Connect
```

```python
df = spark.read.table("population_metrics.default.countries_population")

df.toPandas()
```

📌 Using select() with String Column Names

```python
df.select("COUNTRY_ID", "NAME", "POPULATION").toPandas()
```

```python
df.select("*").toPandas()
```

📌 Using select() with Bracket Notation

```python
df.select(df['country_id'].alias("Id"), df['name'].alias("Country_Name"), df['population']).show()
```

📌 Using select() with DataFrame Attribute Access

```python
df.select(df.COUNTRY_ID , df.NAME, df.POPULATION).toPandas()
```

📌 Using select() with col()

```python
from pyspark.sql.functions import col
#df.select(col("country_id"), col("name"), col("population")).toPandas()
df.select(col("country_id").alias("Id"), col("name").alias("Country_Name"), col("population")).toPandas()
```
