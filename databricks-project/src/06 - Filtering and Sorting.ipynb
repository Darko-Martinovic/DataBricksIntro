```python
import os
if not os.getenv("DATABRICKS_RUNTIME_VERSION"):
    from spark_setup import spark, display  # local run through Databricks Connect
```

📌 Creating a Sample DataFrame

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
from pyspark.sql import functions as F

schema = StructType([
    StructField("employee_id", IntegerType(), True),
    StructField("name",        StringType(),  True),
    StructField("department",  StringType(),  True),
    StructField("salary",      IntegerType(), True),
    StructField("years",       IntegerType(), True)
])

data = [
    (1,  "Alice",   "Engineering", 95000, 5),
    (2,  "Bob",     "Marketing",   62000, 3),
    (3,  "Carol",   "Engineering", 87000, 8),
    (4,  "David",   "HR",          54000, 2),
    (5,  "Eve",     "Marketing",   71000, 6),
    (6,  "Frank",   "Engineering", 110000, 10),
    (7,  "Grace",   "HR",          58000, 4),
    (8,  "Henry",   "Marketing",   67000, 1),
    (9,  "Isla",    "Engineering", 92000, 7),
    (10, "Jack",    "HR",          60000, 3)
]

df = spark.createDataFrame(data, schema=schema)
df.show()
```

📌 filter() — Keep Rows Matching a Condition

```python
# Filter rows where salary is greater than 70000
df.filter(F.col("salary") > 70000).show()
```

📌 where() — SQL-Style String Condition (alias for filter)

```python
# where() is identical to filter() — use whichever reads more naturally
df.where("department = 'Engineering'").show()
```

📌 Chaining Multiple Filters

```python
# Engineering employees with more than 5 years of experience
df.filter(
    (F.col("department") == "Engineering") & (F.col("years") > 5)
).show()
```

📌 orderBy() — Sort Ascending and Descending

```python
# Sort by salary descending
df.orderBy(F.col("salary").desc()).show()
```

📌 sort() — Multi-Column Sort

```python
# Sort by department ascending, then salary descending within each department
df.sort(F.col("department").asc(), F.col("salary").desc()).show()
```

📌 Combining Filter and Sort

```python
# Top earners outside Engineering, sorted by salary descending
df.filter(F.col("department") != "Engineering") \
  .orderBy(F.col("salary").desc()) \
  .show()
```
